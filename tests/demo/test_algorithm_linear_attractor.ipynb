```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

def f(t, state, omega=2.0):
    x, y = state
    r = np.hypot(x, y)
    if r == 0:
        dr = 0.0
        theta = 0.0
    else:
        dr = -r * (r - 1.0) * (r - 2.0) * (r - 3.0)
        theta = np.arctan2(y, x)
    dx = dr * np.cos(theta) - r * np.sin(theta) * omega
    dy = dr * np.sin(theta) + r * np.cos(theta) * omega
    return [dx, dy]

t_span = (0.0, 100.0)
t_eval = np.linspace(*t_span, 5000)

# several initial radii to show attraction/repulsion
inits = [[0.1, 0.0], [0.9, 0.0], [1.5, 0.0], [2.5, 0.0], [4.0, 0.0]]
plt.figure(figsize=(6,6))
for s0 in inits:
    sol = solve_ivp(f, t_span, s0, t_eval=t_eval, rtol=1e-8, atol=1e-10)
    plt.plot(sol.y[0], sol.y[1], lw=0.8)
# draw circles r=1, r=2 (unstable), r=3
theta = np.linspace(0, 2*np.pi, 300)
plt.plot(np.cos(theta)*1.0, np.sin(theta)*1.0, '--', label='r=1 (stable)')
plt.plot(np.cos(theta)*2.0, np.sin(theta)*2.0, ':', label='r=2 (unstable)')
plt.plot(np.cos(theta)*3.0, np.sin(theta)*3.0, '--', label='r=3 (stable)')
plt.gca().set_aspect('equal', 'box')
plt.legend()
plt.xlabel('x'); plt.ylabel('y')
plt.title('Two stable concentric limit cycles (unstable between)')
plt.show()
```
